```python
### 시퀀스 자료형, 배열
# 배열(list)
numbers = [1,2,3,4]
result = numbers[2]
print(result)
# 3

my_ary=[1,"1",True]

#중첩배열, 2차원 배열
nested = [
    [1,2,3],
    [4,5,6],
    [4,8,9]
]
#nested[0][1]
# 2를 꺼낼려면
first_result = nested[0]
result = first_result[1]
print(result)

result = nested[0][1]

fruits = ["사과", "배", "포도", "딸기", "수박"]
#안에 든 거 요소 element(s)
#길이는 값이 몇개가 든 건가
# 배열의길이=5 인덱스(끝값)=4 0부터 시작하니까
fruits[0] #한개 가져온 거=단권조회=인덱스 번호로
#사과
fruits[-1] #수박
#인덱스보다 더 작은 값이나 큰 값을 주면 오류가 난다

#배열 슬라이싱
fruits[1:3] # 배 포도 딸기 수박
print(fruits[1:3]) # ['배', '포도']
print(fruits[0:4:2]) # ['사과', '포도']
print(fruits[::-1]) #['수박', '딸기', '포도', '배', '사과']

#원본배열에 하나를 추가
fruits.append('귤')
print(fruits)
```
