함수는 특정한 목적을 위하여 작성된 코드의 묶음에 이름을 붙여서 재사용할 수 있도록 정의한 것.

'print()'

### 함수의 구조
'''python
def 함수_이름():
    명령문
'''

'''python #함수의 정의부
def 함수_이름(변수1, 변수2, ...): #함수의 시그니처 #변수는 매개(파라미터) 변수
    명령문 #함수의 구현부/함수 본문
'''
매개변수는 함수를 정의할 때 괄호 안에 선언하는 변수 이름입니다. 처음에는 값이 비어 있지만, 함수가 호출되면 전달된 값이 이 변수에 할당

```python
def sum (a,b):
    print(a+b) #표준출력 함수 호출 #매개변수
sum(a=100, b=20)  # 매개변수


def greeting():
    print('Hello, World!')
greeting()

sum(5,3) #인수, 인자
```

```python
def multiply(a,b):
    return a*b #return 뒤에는 값이나 표현식을 쓸 수 있다

def multiply(a,b):
    result =  a*b
    return result #위의 식이나 아래 식이나 같음

multiply(3,3) #곱셈은 했는데 값이 안 나온 것
```

```python
def multiply(a,b):
    return a*b

result = multiply(3,3) #할당해주면 값이 나옴(반환)
print(result)
```
