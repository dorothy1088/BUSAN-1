# series의 이해
series에 저장된 여러 값을 범위로 선택하고 조건에 맞는 값만 추출하며 값을 수정하거나 연산하는 방법

```python
import pandas as pd
import numpy as np
```

```python
sales = pd.Series(
    [120, 95, 143, 88, 176, 132, 101],
    index=['월', '화','수','목', '금','토','일'],
    name='일별_판매량'
)
sales
```

```python
sales_ary = [120,95,143,88,176,132,101]
sales[1:4]
sales.iloc[1:4]
# 위 두개 다 결과값은 동일 
sales.iloc[::-1] #일부터 월까지 일-토-...-월 순으로
sales.iloc[1:5:2] 
```

```python
sales.iloc[3:100] #슬라이싱이라 오류 안나 #또다른 시리즈가 반환
sales.iloc[100] #버그
```

```python
sales.iloc[10:20] #오류 안나 #비어있는 시리즈를 반환
```

```python
sales.loc['화':'목'] #슬라이싱은 범위 기반, 정렬해서 가지고 오니까 다행히 오류 안나
```

```python
sales.loc['월']
sales.loc['수']
sales.loc['금']

a = ['월','수','금']
for item in a:
    print(sales.loc[item]) #시리즈에서 가지고 옴

```

# series 조건 필터링(conditional filtering)
조건에 맞는 데이터들만 필터링해서 가져오는 방법

```python
scores = pd.Series(
    [88,95,62,74,100,55,81],
    index=['학생1', '학생2', '학생3', '학생4', '학생5','학생6', '학생7',],
    name='중간고사 점수'
)
scores
```

## 조건 필터링(boolean series)
score_ary = [88,95,62,74,100,55,81]
score_ary > 70 원래 안되나 시리즈에선 지원해줌
pandas 의 series는 비교연산자의 연산을 지원

```python
scores >= 80
```

```python
scores[scores >= 80] 
```

```python
True and True # 됨
(scores >=80 )and (scores < 90) #비트 연산자 -> 시리즈 : 논리연산자에 넣어
if(true & true) # 비트 논리 연산자 
(scores >=80 ) & (scores < 90) #<-가능
```

```python
(scores >=80 ) & (scores < 90)
(scores >=80 ) | (scores < 90) #앞 소괄호부터 계산
```

```python
scores.loc[(scores >= 80) & (scores < 90)]
```

```python
condition = (scores >= 80) & (scores < 90)
scores.loc[condition]
```

```python
~(scores >= 80) #비트 반전, 내용 바긘 거 
```

```python
scores[~(scores >= 80)]
```

## boolean series를 이용한 집계
python은 ?

```python
scores = pd.Series(
    [88,95,62,74,100,55,81],
    index=['학생1', '학생2', '학생3', '학생4', '학생5','학생6', '학생7',],
    name='중간고사 점수'
)
```

```python
scores.sum()
```

```python
print(f'합격자 수 = {(scores >= 80).sum()}')
print(f'합격률 = {(scores >= 80).mean()}')
print(f'한명이라도 만점이 있는가? = {(scores == 100).any()}') # 어떤 거 하나라도 만족?
print(f'전원이 60점 이상이니? = {(scores >= 60).all()}') #전부 다 만족?

# 시리즈에 달려진 친구들..블린이 아니라
```
