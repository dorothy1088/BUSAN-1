## longform data
일반적으로 하나의 열이 하나의 변수 하나의 행이 하나의 관측값을 갖는 구조

```python
sales_df = pd.DataFrame({
    "advertising": [
        10, 15, 20, 25,
        12, 18, 23, 28,
        8, 14, 19, 24,
    ],
    "sales": [
        120, 155, 190, 230,
        110, 150, 185, 215,
        95, 135, 170, 205,
    ],
    "region": [
        "Seoul", "Seoul", "Seoul", "Seoul",
        "Busan", "Busan", "Busan", "Busan",
        "Daejeon", "Daejeon", "Daejeon", "Daejeon",
    ],
    "visitors": [
        850, 1020, 1180, 1350,
        780, 950, 1100, 1280,
        690, 870, 1030, 1200,
    ],
    "channel": [
        "Online", "Offline", "Online", "Offline",
        "Online", "Offline", "Online", "Offline",
        "Online", "Offline", "Online", "Offline",
    ],
})
long_sales = pd.DataFrame({
    "month": [
        "Jan", "Jan", "Jan",
        "Feb", "Feb", "Feb",
        "Mar", "Mar", "Mar",
        "Apr", "Apr", "Apr",
    ],
    "region": [
        "Seoul", "Busan", "Daejeon",
        "Seoul", "Busan", "Daejeon",
        "Seoul", "Busan", "Daejeon",
        "Seoul", "Busan", "Daejeon",
    ],
    "sales": [
        120, 100, 90,
        150, 130, 115,
        180, 160, 145,
        210, 190, 175,
    ],
})
```

```python
sns.lineplot(
    data = long_sales,
    x='month',
    y='sales',
    hue='region'
)
```

```python
sns.scatterplot(
    data = long_sales,
    x='month',
    y='sales',
    hue='region'

)
```

```python
sns.scatterplot(
    data = long_sales,
    x='month',
    y='sales',
    hue='region'

)
plt.show()
```

## wide form data

```python
sns.lineplot(data=wide_sales)
plt.show()
```
