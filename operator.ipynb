# 연산자
연산자(operator)는 특정한 계산이나 처리를 수행하도록
지시하는 기호 또는 키워드를 의미한다.

연산자에 의해서 처리되는 대상이 되는 값이나 변수를 피연산자(operand)라고 일컫는다.
 
## 연산자의 구분
연산자는 자신이 처리하는 피연산자의 갯수에 따라서 구분이 가능하다.

### 피연산자 갯수에 따른 구분
### 단항연산자(unary operator)
단항연산자는 하나의 피연산자에 대해서 연산을 수행하는 연산자들을 의미한다.
 
### 이항연산자(binray operator)
이항연산자 두개의 피연산자에 대해서 연산을 수행하는 연산자를 의미한다. 

### 다항연산자(ternary operator), 삼항 연산자
다항연산자는 피연산자를 세개 혹은 그 이상을 요구하는 연산자를 의미한다. 있지만 이건 조건식으로 이해할 것.

### 기능에 따른 구분
### 산술 연산자(arithmetric operator)
산술 연산자는 수학적 계산을 수행하는데 사용되는 연산자로서, 한개 혹은 두개의 숫자의 수학적인 작업을 처리하는데 사용한다. 

```python
#a = 10
#b = 20

a,b,c,d,e = 5,10,0.5,True,False

integer_result = a+b
print(integer_result)
#연산은 같은 자료형끼리만 가능
#문자는 배열
# 정수 10(4바이트) 실수 10.5(8바이트) -> 4바이트를 8바이트로 바꾸어 계산은 가능
# 정수랑 실수는 계산 가능 정수가 실수로 바뀜
# 아예 못 바뀌는 것들은 안됨
#그러나 파이썬은 문자+숫자도 가능 숫자를 문자로 바꿈
#정수+실수=실수(정수로 바꿀려면 실수로 나온 걸 강제로 바꾸면 가능) 실수+실수=실수 
#둘 중 큰 것으로 바꿔주는 게 파이썬

float_result= b+c
print(float_result)

integer_result_2= a+d
print(integer_result_2)

#문자열과 문자열을 덧셈=배열과 배열 덧셈도 가능

hello_str="Hello,"
world_Str="World!"
result= hello_str+world_Str
print(result)

list1= [ 1, 2, 3]
list2 = [4,5,6  ]
print(list1+list2)
print(list1*3) # 각 삼을 곱하고 싶으면 넘파이?
```

```python
a, b, c = 5, 10, 0.5
integer_result=b-a
print(integer_result)

integer_result=a-b
print(integer_result)

print(a*b) # **을 쓰면 거듭제곱, *만 쓰면 곱하기
print(a**b)

print("="*50)
print("곱셈연산자")
print("="*50)

```

```python
#모듈러 연산자=나머지 연산자
print(10/4)
#몫 연산자 
print(10//4)

print(10%4)

```

### 비교 연산자
#비교 연산자는 왼쪽 값과 오른 값을 비교하는 거

```python
print(5>3)
print(5<3)
#무조건 왼쪽 기준
print(5>=3)
print(5<=3)
#print(5=3) # 이건 틀림
print(5==3) # 같다는 표시는 두개
print(5!=3) #같지 않다

print(5==5.0)
#둘은 다른 타입이나 값은 같아서 true

x = 5
x>1
x<10

#연쇄비교
1<x<10 #연쇄비교 가능하다

x=5
print(1<x and x<10) #일반 방식
print(1<x<10) #연쇄 비교

```
