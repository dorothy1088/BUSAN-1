```python
student = {
    '수학':90,
    '과학':90,
    '영어':20
}
#반정형 중 제이슨-딕셔너리로 바꿀 수 있는 걸로 바꿔줌
#딕셔너리를 시리즈로 바꿀 수 있음 배열 뿐 아니라
#중괄호({})를 사용해 만든 딕셔너리를 판다스의 pd.Series() 함수에 넣으면 시리즈로 바로 변경.
#이때 딕셔너리의 구성 요소들은 시리즈의 다음과 같은 특징으로 자동 변환됩니다.
#Key (키) → 시리즈의 Index (인덱스) [1]
#Value (값) → 시리즈의 Data (데이터 값) [1]
pd.Series(student)
```

```python
data = {
    '서울': 10,
    '부산': 20,
    '대전':301,
    '천안':235
}
labeled_data = pd.Series(data, index = ['서울', '부산'])
labeled_data

labeled_data = pd.Series(data, index = ['서울', '부산', '대구']) # 대구는 결측치로 Nan
labeled_data
```

```python

#소괄호가 있는 것 = 클래스 내부의 "함수"=메서드(Method)=내부에 저장된 변수들을 가지고 새로운 계산을 하거나 동작을 수행(Action)하는 함수
#함수를 실행하려면 반드시 뒤에 괄호()를 붙여야 동작
#소괄호가 없는 것 = 클래스 내부의 "변수"=속성(Attribute)=메모리에 존재하는 값을 그냥 꺼내서 가져오는(Read) 역할
#시리즈 내 속성(시리즈 점 찍고) 
```

```python
a = 10
def b():
    print(1)
b()

#변수의 유효범위
#a = 10: 전역 변수 a에 정수 10을 할당합니다. (이 코드에서는 이후에 사용되지 않습니다.)
#def b():: b라는 이름의 함수를 정의합니다. 
#print(1): 함수 b가 호출되면 콘솔에 숫자 1을 출력합니다.
#b(): 정의된 함수 b를 실행(호출)하여 화면에 1이 나타납니다.
```
