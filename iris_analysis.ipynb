### 기술통계에서 살펴보는 데이터와 목적
중심경향치(데이터가 어디에 모여 있는가?)
 -평균(mean): 모든 값을 더한 뒤 갯수로 나눈 값
 -중앙값(median): 크기 순서대로 나열했을 때 정 중앙에 있는 값
 -최빈값(mode): 데이터에서 가장 자주 등장하는 값

산포도(데이터가 어떻게 퍼져있는지)
 -분산(variance): 평균에서 데이터가 떨어진 정도를 제곱해서 구한 평균
 -표준편차(standard deviation): 분산의 양의 제곱근
 -사분위수(quartile): 전제 데이터를 4등분한 위치의 값

분포의 형태(데이터의 모양)
 -왜도: 분포가 한쪽으로 치우쳐진 정도
 -첨도: 분포의 뾰족한 정도
