```python
import matplotlib.pyplot as plt
import pandas as pd
df=pd.DataFrame({
    'advertising': [10,15,20,25,30,35,40,45],
    'sales': [100,115,125,145,150,175,185,195]
})

fig, ax= plt.subplots(figsize=(8,6))
ax.scatter(
    df['advertising'],
    df['sales'],
    color='hotpink',
    s=150
)
ax.set_title('Advertising and Sales')
ax.set_xlabel('Advertising')
ax.set_ylabel('Sales')
plt.show()
```

```python
df = pd.DataFrame({
    'study_hours': [2,3,4,5,6,2.5,4.5,5.5,7],
    'scores':[60,67,72,80,86,58,75,82,91],
    'class': ['A','A','A','A','A','B','B','B','B',]
})

df

fig, ax = plt.subplots(figsize=(8, 6))
df[df['class'] == 'A'] # a반의 성적만
class_A=df[df['class'] == 'A']
class_B=df[df['class'] == 'B']

ax.scatter(
    class_A['study_hours'],
    class_A['scores'],
    label='class A'
)

ax.scatter(
    class_B['study_hours'],
    class_B['scores'],
    label='class B'
)

ax.set_xlabel('study_hours')
ax.set_ylabel('scores')

ax.legend()
plt.show()

```

```python
df = pd.DataFrame({
    'x': [1,2,3,2,3,4,3,4,5],
    'y': [5,7,8,4,6,9,3,7,10],
    'group': [
        'A','A','A',
        'B','B','B',
        'C','C','C',
    ]
})


fig, ax = plt.subplots(figsize=(8,6))

group_a= df[df['group']=='A']

ax.scatter(
    group_a['x'],
    group_a['y'],
    label='Group A',
    s=80
)
ax.legend()
plt.show()
```

```python
df = pd.DataFrame({
    'x': [1,2,3,2,3,4,3,4,5],
    'y': [5,7,8,4,6,9,3,7,10],
    'group': [
        'A','A','A',
        'B','B','B',
        'C','C','C',
    ]
})


fig, ax = plt.subplots(figsize=(8,6))

for group_name in ['A','B','C']:
                    group_a= df[df['group']==group_name]

ax.scatter(
    group_a['x'],
    group_a['y'],
    label=f'Group {group_name}',
    s=80
)
ax.legend()
plt.show()
```

```python
import numpy as np
rng = np.random.default_rng(42) #랜덤으로 임의의 수 나옴

x= rng.normal(
    loc=50,
    scale=10,
    size=500
)
x
y=(x * 0.7 + rng.normal(loc=0, scale=8, size=500))

print(len(x),len(y))
```
