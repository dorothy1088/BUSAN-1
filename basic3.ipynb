```python
x= [1,2,3,4,5]
y=[10,15,13,20,24]

fig, axes = plt.subplots(
    figsize=[8,4]
)

axes.plot(
    x,
    y,
    linestyle='dashed',
    marker='o')

plt.show()
```

```python
x= [1,2,3,4,5]
y=[10,15,13,20,24]

fig, axes = plt.subplots(figsize=[8,4])

axes.plot(x, y,
          linestyle='--',
          linewidth=2,
          marker='^')

plt.show()
```

```python
x= [1,2,3,4,5]
y=[10,15,13,20,24]

fig, axes = plt.subplots(
    figsize=[8,4],
)
axes.plot(
    x,
    y,
    linestyle='--',
    linewidth=8,
    marker='^')

plt.show()
```

```python
fig, axes = plt.subplots(figsize=(8,4))
month = [1,2,3,4,5,6]
product_a=[120,130,128,160,175,190]
product_b=[100,110,125,140, 160,165]

axes.plot(month, product_a,marker='o',label='product A')
axes.plot(month, product_b,marker='s', label='product B')

axes.legend() #artist

plt.show()
```
