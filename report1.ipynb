```python
CSV_PATH = Path("./data/부산항만공사_물동량 예측_20250331.csv")
df = pd.read_csv(CSV_PATH, encoding='cp949')
```

```python
df=(df.sort_values(by=["연도","월"],ascending=[True,True]).reset_index(drop=True))
df
```

```python
grouped = df.groupby('연도')
result = grouped['물동량'].mean()
print(result)

#년도마다 물동량이 올라가고 있기에(일부 년도에서 감소한 부분도 있으나) 
#장기적으로 성장하고 있다고 볼 수 있습니다.
```

```python
resul = result.pct_change() * 100
resul

#년도마다 물동량이 올라가고 있기에(일부 년도에서 감소한 부분도 있으나) 
#장기적으로 성장하고 있다고 볼 수 있습니다.
```

```python
resull = pd.DataFrame({
    '연도별 물동량 평균': result,
    '전년 대비 증감률(%)': resul
})

resull['전년 대비 증감률(%)'] = resull['전년 대비 증감률(%)'].round(2)

resull
#년도마다 물동량이 올라가고 있기에(일부 년도에서 감소한 부분도 있으나) 
#장기적으로 성장하고 있다고 볼 수 있습니다.
```

```python
df_filtered = df[~df['연월'].str.startswith('2025')]
result4 = df_filtered.groupby('월')['물동량'].mean()
result4
#2월에만 소폭 감소한 점을 보았을 때 
#일단 바로 예측할 수 있는 것은 2월에는 다른 월보다 날짜일수(28일)가 적기 때문이 아닌가 예상이 됩니다.
```

```python
df_filtered = df[~df['연월'].str.startswith('2025')]
total1 = df_filtered.groupby('연도')['물동량'].sum()
df_trans1 = df_filtered[
    df_filtered['수출입구분설명'].isin(['수출환적', '수입환적'])
]
trans1 = df_trans1.groupby('연도')['물동량'].sum()
ratio1 = (trans1 / total1) * 100
ratio1
#년도별로 비교했을 때 매년 갈수록 환적 비중이 높아지는 것을 확인할 수 있습니다.
#고로 부산항은 환적 기능을 수행하고 있다고 볼 수 있습니다. 
```

```python
df_filtered = df[~df['연월'].str.startswith('2025')]
total2 = df_filtered.groupby('연도')['물동량'].sum()
df_trans2 = df_filtered[
    df_filtered['수출입구분설명'].isin(['수출', '수입'])
]
trans2 = df_trans2.groupby('연도')['물동량'].sum()
ratio2 = (trans2 / total2) * 100
ratio2
```

```python
df_filtered = df[~df['연월'].str.startswith('2025')]
total = df_filtered.groupby('월')['물동량'].sum()
trans = (
    df_filtered[df_filtered['수출입구분설명'].eq('수입환적', '수출환적')]
    .groupby('월')['물동량']
    .sum())
ratio = trans / total * 100
ratio
```
