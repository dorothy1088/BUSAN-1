```python
## if elif else
# else if 와 같이 if절 평가 내용이 거짓일 경우 다른 조건을 추가하고자 할 때 사용

score = 78
grade = "F"

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
elif score >= 60:
    grade = "D"
else:
    grade = "F"
print(grade)

# 첫번째에서 참이라고 평가되면 나머지는 평가안함
```

```python
## 반복문 (LOOP Statement)

#반복문은 특정 코드 블록을 조건이 만족되는 동안 또는 지정된 횟수만큼 실행하는 제어구조.

print('Python은 너무너무 재밌어. 최고야')

print('안녕하세요')

#while 
# while 문은 주어진 조건식이 참인 동안에 코드블록을 반복해서 실행하는 제어구조

#-while문 구조
'''python
while 조건식 or 값:
    반복할_명령문
'''

#is_python_easy = True
while is_python_easy:
    print('계속 공부해야지')
```

```python
count = 1 #초기식-이게 무한 반복
while count <= 5: # 조건식
    print(f'{count}번 반복해서 출력합니다!')
    count = count + 1 #무한 반복 피하기 위해, 증감식
print('완료!')

#돌아갈 포인트를 미리 깃발 뽑기
#스텝 메모리-돌아갈 포인트
#스탭 오버플로우 


#제어문은 초기식, 조건식, 증감식이 필요함
```

```python
n=5
while n>0:
    print(n, end=" ") # 보통은 그냥 개행인데 " " 설정함으로 옆으로 나열 및 간격을 맞춰줌 ,나 슬러쉬 등도 활용 가능
    n=n-1
print('발사!')
```

```python
#반복문의 제어를 조정하는 것
#분기문(branching statement)
#분기문은 명령의 흐름을 제어하여 특정 코드블록으로 이동하거나 반복문을 종료하는 등 이전의 제어문들과
#마찬가지로 명령 흐름의 분기를 생성할 수 있는 문법

## break
#break는 현재 실행중인 루프(반복)를 즉시 탈출하는 키워드

for i in range(10):
    if i == 5 :
        print(f'{i}에서 break가 발동!')
        break
    print(f'현재 숫자={i}')

```

```python
##continue
#현재 반복의 나머지 코드를 건너 뛰고 다음 반복으로 즉시 이동하는 키워드

for i in range(1,11): #항상 끝점은 -1이기에 지금 이것은 1부터 10까지다
    print(f'이번 손님 = {i}')
    if i % 2 == 0:
        continue #다음 손님 버튼, 반복문 종료가 아니라 순서를 건너뛰는 거
    print(i,end=' ')


for i in range(1,11): #항상 끝점은 -1이기에 지금 이것은 1부터 10까지다
    print(f'이번 손님 = {i}')
    if i % 2 == 0:
        continue #다음 손님 버튼, 반복문 종료가 아니라 순서를 건너뛰는 거
    print(i)

```
