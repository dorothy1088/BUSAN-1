```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
from seaborn import palettes
```

```python
sales_df = pd.DataFrame({
    "month": [
        1, 2, 3, 4, 5, 6,
        1, 2, 3, 4, 5, 6,
        1, 2, 3, 4, 5, 6,
    ],
    "region": [
        "Seoul", "Seoul", "Seoul",
        "Seoul", "Seoul", "Seoul",
        "Busan", "Busan", "Busan",
        "Busan", "Busan", "Busan",
        "Daejeon", "Daejeon", "Daejeon",
        "Daejeon", "Daejeon", "Daejeon",
    ],
    "sales": [
        120, 138, 155, 171, 188, 205,
        105, 118, 132, 149, 163, 178,
        92, 106, 119, 133, 148, 162,
    ],
    "advertising": [
        10, 12, 15, 18, 21, 24,
        9, 11, 13, 16, 19, 22,
        7, 9, 12, 14, 17, 20,
    ],
    "visitors": [
        800, 900, 1010, 1130, 1260, 1400,
        720, 810, 900, 1010, 1120, 1250,
        650, 730, 820, 910, 1020, 1140,
    ],
})
sales_df
```

```python
sns.set_theme()
```

```python
sns.lineplot(
    data=sales_df,
    x="month",
    y="sales",
    hue="region",
    marker='o'
)
```

# color palette
1. 서로 다른 범주 구분
    - 범주형 팔레트: 서로 순서가 없는 데이터를
2. 작은 값에서 큰 값으로 변화 표현
    -연속형 팔레트
3. 특정 기준보다 낮거나 높은 방향 표현
    -발산형 팔레트

```python
df= pd.DataFrame([
    [10,20,30],
    [20,40,60],
    [30,60,90],
],
    index=['A','B','C'],
    columns=['X','Y','Z']
)
```

```python
sns.heatmap(
    data=df,
    annot=True,
    cmap='coolwarm',
    cbar=True
)
```
