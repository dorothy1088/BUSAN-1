```python
import matplotlib.pyplot as plt
```

```python
products = ['A', 'B', 'C', 'D']
sales = [120,180,150,210]

fig, ax = plt.subplots(figsize=(10,4))
ax.bar(products, sales) # 가로, 세로
plt.show()
```

```python
products = ['product A', 'product B', 'product C', 'product D']
sales = [120,180,150,210]

fig, ax = plt.subplots(figsize=(8,4))

ax.barh(products, sales)

ax.set_title('products sales')
ax.set_xlabel('sales')
ax.set_ylabel('products')

plt.show()
```

```python
products = ['product A', 'product B', 'product C', 'product D']
sales = [120,180,150,210]

fig, ax = plt.subplots(figsize=(10,4))

bars = ax.barh(products, sales)
ax.bar_label(bars,
             padding=3)

ax.set_title('products sales')
ax.set_xlabel('sales')
ax.set_ylabel('products')

plt.show()
```

```python
import numpy as np

products = ['A', 'B', 'C']
sales_2026 = [120,160,140]
sales_2025 = [150,180,170]

x = np.arange(len(products)) # 숫자배열
width = 0.2

fig, ax = plt.subplots(figsize=(8,4))
ax.bar(
    x-width/2,
    sales_2025,
    width,
    label='2025'
)

ax.bar(
    x+width/2,
    sales_2026,
    width,
    label='2026'
)

ax.set_xticks(x)
ax.set_xticklabels(products)


ax.set_title('sales by year')
ax.set_xlabel('products')
ax.set_ylabel('sales')

ax.legend()
plt.show()
```

```python
products = ['A', 'B', 'C']
q1 = [120,150,130]
q2 = [140,170,145]
q3 = [160,180,155]

x= np.arange(len(products))
width = 0.25

fig, ax = plt.subplots(figsize=(10,4))

ax.bar(x-width,
       q1,
       width=width,
       label='Q1')

ax.bar(x,
       q2,
       width=width,
       label='Q2')

ax.bar(x+width,
       q3,
       width=width,
       label='Q3')


ax.set_xticks(x)
ax.set_xticklabels(products)

ax.legend()

plt.show()
```

```python
products = ['A', 'B', 'C']
online = [100,130,130]
offline = [50,70,80]

fig, ax = plt.subplots(figsize=(8,4))
ax.bar(
    products,
    online,
    label='Online'
)
ax.bar(
    products,
    offline,
    bottom = online,
    label='Offline'
)
ax.bar(
    products,
    [10,20,30],
    label='Others'
)

ax.legend()
plt.show()
```
